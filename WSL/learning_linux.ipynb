# Linux 

** *
## Basics

Linux is `different` from Windows.

Everything is `file-based`. Even devices and processes are represented as files.

### File system Root (`/`)

Like `C:\` in Windows, `/` is the `root` of the Linux file system. Common folders:

- /home/<username> → your personal files (like C:\Users\You)
- /etc → configuration files
- /usr → installed programs
- /var → logs, cache, variable data

### Case Sensitive

File.txt ≠ file.txt.

### Permissions

Files and folders have owner, group, and permission bits (read/write/execute).
** * 
## Must-Know commands (Everyday Use)

### Navigation

- pwd --> print current working directory
- ls --> list files in current directory
- ls -la --> detailed list with hidden files
- cd <path> ---> change directory

### File & Directory Operations

- touch file.txt ---> create empty file
- mkdir myfolder ---> create a new directory
- cp file.txt copy.txt ---> copy file
- mv file.txt new.txt ---> rename/move file
- rm file.txt ---> remove file
- rm -r folder/ ---> remove folder and contents

### Viewing files

- cat file.txt ---> print entire files
- less file.txt ---> view file with scrolling
- head file.txt ---> first 10 lines
- tail file.txt ---> last 10 lines
- tail -f log.txt ---> follow a log file live

### System & packages

- whoami ---> current user
- uname -a ---> system info
- df -h ---> disk space usage
- top ---> running processes
- htop ---> better than top, may need to be installed

### Package Management (Ubuntu)

- sudo apt update ---> refresh package list
- sudo apt upgrade ---> upgrade installed packages
- sudp apt install <pkg> ---> install new package

### Searching & Finding

- grep "word" file.txt ---> search for word in file
- grep -r "word" folder/ ---> search recursively in folder
- find . -name "*.py" ---> find all python files

### Permissions

- chmod +x script.sh ---> make script executable
- ls -l ---> see file permissions

** *

## Useful shortcuts in the shell

- Ctrl + C → stop running command

- Ctrl + L → clear screen

- Tab → autocomplete file/folder names

- ↑ ↓ → cycle through command history

- !! → repeat last command

- !<string> → run last command starting with <string> (e.g. !python)
