# WSL (Windows Subsystem for Linux)

WSL stands for Windows Subsystem for Linux.

It’s like having Linux running inside your `Windows` computer, without needing to install Linux separately or use a virtual machine.

Think of it as a `Linux mode` you can turn on inside Windows.

## Why to use it for software development?

Most professional developers (especially in backend, cloud, and devops) work on Linux or macOS, because:

- Servers on the internet usually run Linux.
- Many programming tools (Python, Node, Docker, etc.) are easier to run on Linux.
- Command line tools (like git, ssh, etc.) are more powerful and standard in Linux.

By using `WSL`:

- You get `Linux tools` without leaving Windows.
- You can practice exactly what your team uses (Linux/macOS commands).
- You avoid "it works on my machine but not on the server" problems.
- You can connect it with VS Code, so you still code `comfortably on Windows` but run everything in Linux.

## Example

If you install Python on Windows, some libraries may not work the same way as on Linux.
If you install Python inside WSL, it’s exactly the same environment as your team/server.

## In short

- Windows = good for daily use, Office, games, etc.
- WSL = Linux inside Windows, for development like the pros do it.
