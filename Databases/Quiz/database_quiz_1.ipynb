## What is normalization in databases?

- a) The process of backing up data
- b) Reducing redundancy and improving data integrity
- c) Encrypting data for security

## Solution

b) Reducing redundancy and improving data integrity

### Explanation:

Normalization organizes data to reduce duplication and ensure consistency.

##  What is the purpose of a foreign key in a relational database?

- a) To uniquely identify a row in its own table
- b) To link records between two tables
- c) To allow duplicate rows

## Solution

b) To link records between two tables

### Explanation

 A foreign key establishes a relationship between rows in two tables.

## Which SQL command is used to remove a table and all of its data?

- a) DELETE
- b) REMOVE
- c) DROP

## Solution

c) DROP

### Explanation

DROP TABLE permanently deletes the table and its structure
