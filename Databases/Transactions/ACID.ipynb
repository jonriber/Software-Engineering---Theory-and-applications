# ACID Transactions - Database

ACID is an acronym that stands for `Atomicity`, `Consistency`, `Isolation` and `Durability`.

They ensure reliable processing of transactions in a database system.

## Atomocity

"All or nothing"

A transaction needs to be all or nothing, which means, either every operation within the transaction is completed successfully
or none are.

If any part of the transaction fails, the entire transaction is rolled back, leaving the database unchanged.

### Example

Transferring money from A to B:
- Debit account A
- Debit account B

if the credit fails, debit should also be undone.

## Consistency

"valid data before and after the transaction"

A transaction must bring the database from one valid state to another, maintaining all predefined rules, such as 
constraints, cascades, triggers and foreign keys.

### Example

If there is a constraint that account balances can't go below zero, a transaction taht violates this rule must be 
rejected.

## Isolation

"Transactions don't interfere with each other".

Concurrent transactions should not affect each other's execution. The intermediate state of a transaction must not be 
visible to other transactions.

### Example

If two users are looking the last seat on a flight at the same time, the database must ensure only one booking goes 
through.

## Durability

"Once committed, it's permanent."

Once a transaction is committed, it will survive any subsequent failures (powe loss, crash). Changes are permanently
written to storage.

### Example

If you complete a bank transfer and the server crashes a second later, the transaction will still be in the database
after recovery.
