# Quiz 2 - OOP

# What is the main purpose of using constructors in a class?

- a) To inherit methods from the superclass
- b) To define private variables
- c) To initialize objects when they are created

## Solution

c) To initialize objects when they are created
Explanation: A constructor sets up the initial state of an object (e.g., assigning default values or parameters).


# Which of the following is true about method overriding?

- a) The overridden method must have a different name
- b) The method signature must remain the same
- c) It can only happen in the same class

## Solution

b) The method signature must remain the same
Explanation: Overriding requires the same method name, return type, and parameters in the subclass.

# What is the output of dynamic (runtime) polymorphism dependent on?

- a) The object’s reference type
- b) The compile-time type of the method
- c) The actual object that is referenced

## Solution

c) The actual object that is referenced
Explanation: Runtime polymorphism chooses the method to execute based on the actual object (not the reference type).

# Which of the following best demonstrates encapsulation?

- a) Making all class variables public
- b) Using getters and setters for private variables
- c) Inheriting from another class

## Solution

b) Using getters and setters for private variables
Explanation: Encapsulation involves hiding internal data and exposing access via public methods.

# What keyword is used in many OOP languages to prevent a method from being overridden?

- a) abstract
- b) final
- c) private

## Solution

b) final
Explanation: The final keyword (e.g., in Java) or similar keywords in other languages prevent overriding.

# Which principle of OOP is violated if a class has too many responsibilities?

- a) Polymorphism
- b) Encapsulation
- c) Single Responsibility Principle

## Solution

c) Single Responsibility Principle
Explanation: Each class should have only one responsibility or reason to change.

# What is the difference between an abstract class and an interface (in most OOP languages like Java or C#)?

- a) Abstract classes can’t have constructors, interfaces can
- b) Interfaces can define behavior, abstract classes cannot
- c) Abstract classes can contain implemented methods; interfaces usually can't (in some languages they can)

## Solution

c) Abstract classes can contain implemented methods; interfaces usually can't (in some languages they can)
Explanation: Abstract classes can have method implementations, whereas traditional interfaces only define method signatures.

# What is a benefit of using composition over inheritance?

- a) It reduces code duplication by inheriting functionality
- b) It allows more flexible code reuse by combining behavior
- c) It allows access to private members of other classes

## Solution

b) It allows more flexible code reuse by combining behavior
Explanation: Composition lets you build complex behavior by assembling simpler objects, promoting flexibility.
