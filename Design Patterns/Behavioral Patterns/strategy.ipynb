# Strategy Design Pattern

This is a behavioral design pattern that enables selecting an algorithm behavior at runtime.

It defines a family of algorithms, encapsulates each one and makes them interchangeable. This pattern promotes 
`open/closed principle` from SOLID. Open for extension, closed for modification.

## When to use

- One has many related classes that differ only in their behavior
- one wants to avoid using many if/elif statements to select behavior
- one wants to make code flexible and reusable

```python
# Basic example using python

from abc import ABC, abstractmethod

# Strategy Interface

class PaymentStrategy(ABC):
    @abstractmethod
    def pay(self, amount: float) -> None:
        pass
    
# Concrete Strategies
class CreditCardPayment(PaymentStrategy):
    
    def __init__(self, card_number: str):
        self.card_number = card_number
        
    def pay(self, amount:float) -> None:
        print(f"Paid {amount} using Credit Card {self.card_number}")
    
class PayPalPayment(PaymentStrategy):
    
    def __init__(self, email:str):
        self.email = email
    
    def pay(self, amount:float)-> None:
        print(f"Paid {amount} using PayPal account {self.email}")
        
# Context Class

class ShoppingCart:
    def __init__(self, strategy: PaymentStrategy):
        self.strategy = strategy
        self.items = []
        
    def add_item(self, item: str, price:float):
        self.items.append((item, price))
        
    def checkout(self):
        total = sum(price for _, price in self.items)
        self.strategy.pay(total)
```
