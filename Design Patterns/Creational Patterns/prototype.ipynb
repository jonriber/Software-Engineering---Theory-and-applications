# Prototype - Creational Design Pattern

## What is?

Prototype Creational Design Pattern is a design pattern that allows me to create new objects by `cloning` existing ones (prototypes),
instead of creating them from scratch using constructors. It is very useful when object creation is costly (time or complexity) and I want to avoid
`repeatedly` creating similar objects.

The core Idea is: `Instead of instantiating new objects directly, clone a prototype.`

## Use cases

- If I need to create many similar objects
- Object creation is very expensive (deep configurations, I/O)
- I want to keep copies of objects at various states

```python
# Python example of prototype creational design pattern

import copy

class Prototype:
    def clone(self):
        return copy.deepcopy(self)
    
# Concrete prototype classes

class Car(Prototype):
    
    def __init__(self, brand, model, options):
        self.brand = brand
        self.model = model
        self.options = options
        
    def __str__(self) -> str:
        return f"{self.brand} {self.model} with options: {', '.join(self.options)}"
    
# Usage Example

if __name__ == "__main__":
    base_car = Car("Toyota", "Camry", ["Sunroof", "Leather Seats"])
    car1 = base_car.clone()
    car1.options.append("Navigation System")
    car2 = base_car.clone()
    car2.options.append("Backup Camera")
```

## When to use

| Situation                          | Use Prototype? |
| ---------------------------------- | -------------- |
| Creating an object is slow/complex | ✅ Yes          |
| Many variations of the same object | ✅ Yes          |
| Few objects to reuse               | ❌ No           |


## Pitfalls

- Be careful with shallow vs deep copies (`copy.copy` vs `copy.deepcopy`)
- Cloning may violate `encapsulation` if internals are exposed
- Managing cloneable state can get `messy in mutable nested objects`

## Best Practices

- Implement `.clone()` method explicitly
- Consider using `copy.deepcopy` unless I am sure a shallow copy is enough
- Use with configuration heavy or templated objects

## Real World Analogy

Use case: Creating ID Badges at a conference

Image I need to organize a tech conference. Therefore, I need to create hundreds of ID Badges for attendees.

`Without` Prototyping I would need to manually set up each badge:

- Add conference Logo
- Add background design
- set font and color
- add attendee name
- add QR code

Using this setup everytime from scratch is tedious and error-likely.

`with` Prototyping, I would need to create a template badge (prototype) with the common stuff already in place:

- Logo, design, font, default layout

Now, for each new attendee, I need to:

- Clone the badge template
- Customize the name and QR code
- Print

This is the prototype creational pattern in action.

```python

# Real World Example

import copy

class Badge:
    
    def __init__(self, template, name=None, qr_code=None):
        self.template = template
        self.name = name
        self.qr_code = qr_code
        
    def clone(self):
        return copy.deepcopy(self)
    
    def __str__(self):
        return f"{self.template} Name: {self.name}, QR Code: {self.qr_code}"
    
    
# Usage Example

prototype_badge = Badge(template="DevCon 2025 - VIP Layout")

badge_1 = prototype_badge.clone()
badge_1.name = "Alice"
badge_1.qr_code = "QR12345"

badge_2 = prototype_badge.clone()
badge_2.name = "Bob"
badge_2.qr_code = "QR67890"

print(badge_1)
print(badge_2)
```
