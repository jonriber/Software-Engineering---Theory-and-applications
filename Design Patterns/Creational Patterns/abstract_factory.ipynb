# Abstract Factory - Creational Design Pattern

The `Abstract Factory` design pattern is a creational pattern that provides an `interface` for creating families of related
or dependent objects without specifying their concrete classes.

## Core Idea

If I am building a UI toolkit that supports both windows and macOS. Each OS has different implementations for components
like buttons and checkboxes.

This pattern lets me create:
- WindowsButton, WindowsCheckBox
- MacButton, MacCheckBox

All of that without letting my client code needing to know the difference.

## Componentes of the Pattern

- Abstract Factory: Declares a set of methods for creating abstract products
- Concrete Factory: Implements the creation of specific products
- Abstract Product: Declares an `interface` for a type of product
- Concrete Product: `Implements` the abstract product interface
- Client: uses only the abstract interfaces (factories and products)

```python
# ABSTRACT PRODUCTS

from abc import ABC, abstractmethod

class Button(ABC):
  @abstractmethod
  def render(self):
    pass

class Checkbox(ABC):
  @abstractmethod
  def render(self):
    pass
```

```python
# CONCRETE PRODUCTS

class WindowsButton(Button):
  def render(self) -> str:
    return "Rendering a Windows button"
  
class MacButton(Button):
  def render(self) -> str:
    return "Rendering a Mac button"
  
class WindowsCheckbox(Checkbox):
  def render(self) -> str:
    return "Rendering a Windows checkbox"
  
class MacCheckbox(Checkbox):
  def render(self) -> str:
    return "Rendering a Mac checkbox"
```
