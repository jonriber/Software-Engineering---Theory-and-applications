# Bridge - Structural Design Pattern

This is a practical Structural Design Pattern when it clicks and it starts to make sense.

Bridge decouples an `abstraction` from its implementation so it is possible to evolve independently. Instead of subclassing for every combination,
one compose the abstraction with a pluggable implementation.

- Abstraction: The API my app uses
- Implementator: the capability that can vary (platform, backend, algorithm, IO)

## When to use it

- when the abstraction and the implementation to change on different axes is expected
- when swapping implementations at runtime is desired (different backend, local, cloud, mock)
- Fighting a class explosion (BigButton, SmallButton, BigIconButton)



```python
# Python example

from abc import ABC, abstractmethod

class Renderer(ABC):
    @abstractmethod
    def draw_circle(self, x:float, y: float, radius: float) -> None: ...
        
class VectorRenderer(Renderer):
    def draw_circle(self, x, y, radius):
        print(f"Raster: plot pixels for circle at ({x}, {y}) with r={radius}")
        
class Shape(ABC):
    def __init__(self, renderer: Renderer):
        self.renderer = renderer

    @abstractmethod
    def draw(self) -> None: ...

class Circle(Shape):
    def __init__(self, renderer: Renderer, x: float, y: float, r: float):
        super().__init__(renderer)
        self.x, self.y, self.r = x, y, r
    
    def draw(self):
        self.renderer.draw_circle(self.x, self.y, self.r)
        
vector = VectorRenderer()
raster = VectorRenderer()

c1 = Circle(vector, 10, 10, 5)
c2 = Circle(raster, 2,3,7)

c1.draw()  # Raster: plot pixels for circle at (10, 10) with r=5
c2.draw()  # Raster: plot pixels for circle at (2, 3) with r=7
```
