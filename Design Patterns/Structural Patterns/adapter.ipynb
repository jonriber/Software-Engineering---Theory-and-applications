# Adapter - Structural Design Pattern

This pattern allows objects with incompatible interfaces to work together. It acts like a bridge
between two incompatible interfaces.

Image that we have an existing class with a specific interface, and a new class or library
uses a different one.

Instead of rewriting the whole code base or the new class, one can use an adapter to make 
them work together.

## Structure

- Target: the interface expected by the client code
- Adaptee: the existing class with a different interface
- Adapter: translates the target interface to the adaptee

## Real life analogy

Power plug adapter:

- a laptop charger (UK plug) but my wall socket is EU type
- I buy an adapter that fits into EU socket and accepts UK plug on the other side
- Adapter do not change the laptop or the wall, just connect them both.

## WHen to use it

An adapter is useful when:

- I want to reuse an existing class whose interface is incompatible with the rest of the system
- I can't or shouldn't modify the original class
- I want to work with multiple third-party APIs without rewriting my main code

```python
# Python example

import json
import yaml

class JSONConfig:
    def read_json(self, json_string:str):
        return json.loads(json_string)
    
class YAMLConfig:
    def read_yaml(self, yaml_string: str):
        return yaml.safe_load(yaml_string)
    
class YAMLConfigAdapter(JSONConfig):
    def __init__(self, yaml_config: YAMLConfig):
        self.yaml_config = yaml_config
    
    def read_json(self, json_string:str):
        
        data = super().read_json(json_string)
        yaml_string = yaml.dump(data)
        return self.yaml_config.read_yaml(yaml_string)
    
    
old_config_reader = JSONConfig()
yaml_config_reader = YAMLConfigAdapter(YAMLConfig())

# Old config reading (JSON)
json_data = old_config_reader.read_json('{"name": "Alice"}')
print(json_data)

# New config reading via adapter
yaml_data = yaml_config_reader.read_json('{"name": "Bob"}')
print(yaml_data)
```

## Key points for python language

In python, adapter pattern can be implemented usingÇ

- Inheritance (class-based adapter)
- Composition(Object-based adapter) -> wrapping adaptee inside
- Duck typing makes adapters simpler because strict interface definitions is not needed
- Adapters help with backward compatibility during migrations

## Real world use cases

- Wrapping `different database` drivers under a common interface
- Bridging old APIs to new APIs in legacy code
- making third-party SDKs compatible with the application interface
- Unifying data access from multiple formats
