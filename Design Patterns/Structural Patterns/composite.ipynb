# Composite - Structural Design Pattern

This pattern is all about treating individual objects and groups of objects in the same `way`.

It allows to:

- Built tree-like objects structures
- work with leaves (single elements) and composites (collections of elements) using a common interface
- Adding new components without modifying or changing the client node

Practical thinking:

- A folder can contain files or other folders. From the perspective of the file system, both are just simnple components.

## When to use it

- I have hierarchical structure
- I want uniform operations across all nodes
- I want recursive compositions of objects

`Common examples`:

- File systems (files and directories)
- UI frameworks (buttons, panels, windows)
- HTML DOM (elements inside elements)
- Organization charts (employees and managers)

## Structure
```plaintext
Component (interface/abstract)
 ├── Leaf (single object)
 └── Composite (collection of Components)
```

`Participants`:

1. Component — common interface for all objects in the composition.
2. Leaf — represents single objects (end nodes).
3. Composite — represents nodes with children (stores components, delegates work).


````python
# Example in python

from abc import ABC, abstractmethod

class FileSystemItem(ABC):
    @abstractmethod
    def show(self, indent=0):
        pass

class File(FileSystemItem):

    def __init__(self, name):
        self.name = name

    def show(self, indent=0):
        print(" " * indent + f"File: {self.name}")

class Directory(FileSystemItem):

    def __init__(self, name):
        self.name = name
        self.children = []

    def add(self, item: FileSystemItem):
        self.children.append(item)

    def show(self, indent=0):
        print(" " * indent + f"Dir: {self.name}")
        for child in self.children:
            child.show(indent + 2)
# Example usage

root = Directory("root")
file1 = File("file1.txt")
file2 = File("file2.txt")
subdir = Directory("subdir")
subfile = File("subfile.txt")

root.add(file1)
root.add(file2)
root.add(subdir)
subdir.add(subfile)

root.show()
````
